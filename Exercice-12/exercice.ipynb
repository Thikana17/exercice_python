```python
import matplotlib.pyplot as plt
import seaborn as sns


titanic = pd.read_csv("../Exercice-08/data/titanic_propre.csv")

taxis = sns.load_dataset("taxis")
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.histplot(data=titanic, x="age", bins=30, ax=ax)

ax.set_title("La majorité des passagers avaient entre 20 et 30 ans")
ax.set_xlabel("Âge (années)")
ax.set_ylabel("Nombre de passagers")

fig.savefig("figures/1_ages.png", dpi=150, bbox_inches="tight")
plt.show()
```

Nous voulons montrer la distribution d'une variable quantitative (l'âge), le bon graphique est donc un histogramme (histplot)

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=titanic, x="class", y="survived", hue="sex", ax=ax)

ax.set_title("Les femmes de 1re classe ont le taux de survie le plus élevé")
ax.set_xlabel("Classe du billet")
ax.set_ylabel("Taux de survie")

fig.savefig("figures/2_survie_classe_sexe.png", dpi=150, bbox_inches="tight")
plt.show()
```

Nous croisons une mesure (taux de survie) avec des catégories (classe et sexe). Un diagramme en barres (barplot) avec la couleur (hue) pour le sexe est adapté.

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=titanic, x="class", y="survived", hue="sex", ax=ax)

ax.set_title("Les femmes de 1re classe ont le taux de survie le plus élevé")
ax.set_xlabel("Classe du billet")
ax.set_ylabel("Taux de survie")

fig.savefig("figures/2_survie_classe_sexe.png", dpi=150, bbox_inches="tight")
plt.show()
```

Pour montrer la répartition d'une mesure (prix) par catégorie (classe), la boîte à moustaches (boxplot) est adapté.

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.boxplot(data=titanic, x="class", y="fare", ax=ax)

ax.set_title("Les billets de 1re classe sont bien plus chers et dispersés")
ax.set_xlabel("Classe du billet")
ax.set_ylabel("Prix du billet (£)")

fig.savefig("figures/3_prix_classe.png", dpi=150, bbox_inches="tight")
plt.show()
```

Pour montrer une évolution dans le temps (temps vs quantitatif), on utilise une courbe (lineplot).

```python
ca_par_jour = taxis.set_index("pickup").resample("D")["total"].sum().reset_index()

fig, ax = plt.subplots(figsize=(12, 5))
sns.lineplot(data=ca_par_jour, x="pickup", y="total", ax=ax)

ax.set_title("Le chiffre d'affaires des taxis varie selon les jours")
ax.set_xlabel("Date de la course")
ax.set_ylabel("Chiffre d'affaires total ($)")

fig.savefig("figures/4_taxis_ca_jour.png", dpi=150, bbox_inches="tight")
plt.show()
```

Pour montrer le lien entre deux mesures quantitatives (distance et prix), il faut un nuage de points (scatterplot).

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.scatterplot(data=taxis, x="distance", y="total", ax=ax)

ax.set_title("Le prix de la course augmente logiquement avec la distance")
ax.set_xlabel("Distance parcourue (miles)")
ax.set_ylabel("Prix total ($)")

fig.savefig("figures/5_taxis_distance_prix.png", dpi=150, bbox_inches="tight")
plt.show()
```
