```python
import pandas as pd

titanic_propre = pd.read_csv("../Exercice-08/data/titanic_propre.csv")

ports = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
    "pays": ["Angleterre", "France", "Irlande"],
})
ports
```

```python
nb_avant = len(titanic_propre)

titanic_enrichi = titanic_propre.merge(ports, on="embark_town", how="left", validate="many_to_one")

assert len(titanic_enrichi) == nb_avant
print(f"Nombre de lignes après jointure : {len(titanic_enrichi)}")
```

```python
resultats_pays = titanic_enrichi.groupby("pays").agg(
    passagers=("survived", "size"),
    taux_survie=("survived", "mean")
).round(2)

resultats_pays
```

```python
ports_reduit = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg"],
    "pays": ["Angleterre", "France"],
})

test_inner = titanic_propre.merge(ports_reduit, on="embark_town", how="inner")
print(f"Lignes avec inner : {len(test_inner)}")

test_left = titanic_propre.merge(ports_reduit, on="embark_town", how="left")
print(f"Lignes avec left : {len(test_left)}")
```

Avec how="inner", seules les clés présentes des deux côtés sont gardées : nous perdons donc les lignes des passagers ayant embarqué à Queenstown.
Avec how="left", toutes les lignes de la table de gauche (le Titanic) sont gardées : les passagers de Queenstown sont toujours présents, mais auront une valeur manquante (`NaN`) pour la colonne `pays`.
