```python
premiere = df[df["pclass"] == 1]
part_femmes_1re = (premiere["sex"] == "female").mean() * 100
print("Femmes en 1re classe :", round(part_femmes_1re, 1), "%")
```

```python

df[(df["pclass"] == 1) & (df["sex"] == "female")]["survived"].mean() * 100
```

```python
survie_femmes_1re = df[(df["pclass"] == 1) & (df["sex"] == "female")]["survived"].mean() * 100
survie_hommes_3e = df[(df["pclass"] == 3) & (df["sex"] == "male")]["survived"].mean() * 100
print("Survie femmes 1re :", round(survie_femmes_1re, 1), "%")
print("Survie hommes 3e :", round(survie_hommes_3e, 1), "%")
```
