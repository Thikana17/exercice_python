```python
mediane_age = df.groupby(["pclass", "sex"])["age"].transform("median")
df["age"] = df["age"].fillna(mediane_age)
print(mediane_age)
```

```python
df = df.dropna(subset=["embark_town"])
```

```python
df = df.drop(columns=["deck"])
```

```python
df.isna().sum()
```

```python
nb_doublons = df.duplicated().sum()
df[df.duplicated(keep=False)].sort_values(list(df.columns)).head(20)
```

```python
df = df.drop(columns=["embarked", "alive", "adult_male", "who"])

df["sex"] = df["sex"].astype("category")
df["class"] = df["class"].astype("category")
df["survived"] = df["survived"].astype(bool)
```

```python
q1 = df["fare"].quantile(0.25)
q3 = df["fare"].quantile(0.75)
ecart = q3 - q1
borne_haute = q3 + 1.5 * ecart

tres_chers = df[df["fare"] > borne_haute]
len(tres_chers)
```

```python
import os
os.makedirs("data", exist_ok=True)
df.to_csv("data/titanic_propre.csv", index=False)
```
