```python
import pandas as pd
import seaborn as sns

taxis = sns.load_dataset("taxis")
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
```

```python
stats_par_jour = (
    taxis.set_index("pickup")
    .resample("D")
    .agg(
        nombre_courses=("total", "size"),
        chiffre_affaires=("total", "sum")
    )
)
print(stats_par_jour.head())
```

```python
taxis["heure"] = taxis["pickup"].dt.hour
heure_chargee = taxis["heure"].value_counts().head(1)

print(f"L'heure la plus chargée : \n{heure_chargee}")
```

```python
taxis["jour_semaine"] = taxis["pickup"].dt.day_name()
pourboire_moyen_jour = taxis.groupby("jour_semaine")["tip"].mean().sort_values(ascending=False)

print(f"Pourboire moyen par jour : \n{pourboire_moyen_jour}")
```

```python
taxis["duree_minutes"] = (taxis["dropoff"] - taxis["pickup"]).dt.total_seconds() / 60

aberrantes = taxis[(taxis["duree_minutes"] <= 0) | (taxis["duree_minutes"] > 180)]
print(f"Nombre de courses aberrantes : {len(aberrantes)}")

taxis = taxis[(taxis["duree_minutes"] > 0) & (taxis["duree_minutes"] <= 180)]
```
